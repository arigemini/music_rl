```python
import tensorflow.compat.v1 as tf
import magenta
import music_rl
```

```python
tf.disable_v2_behavior()

OUTPUT_DIR = "/Users/asaha/Documents/Stanford/projects/music_rl_output/baseline_note_rnn"
```

```python
hparams = magenta.contrib.training.HParams(
    random_action_probability=0.1,
    store_every_nth=1,
    train_every_nth=5,
    minibatch_size=32,
    discount_rate=0.5,
    max_experience=100000,
    target_network_update_rate=0.01
)

rl_net = music_rl.MusicRl(
    output_dir=OUTPUT_DIR, 
    dqn_hparams=hparams,
    reward_mode='music_theory_all',
    reward_scaler=1,
    output_every_nth=50000,
    num_notes_in_melody=32
)
```

```python
rl_net.generate_music_sequence(visualize_probs=True, title='pre_rl', length=32)
```

```python
stat_dict = rl_net.evaluate_music_theory_metrics(num_compositions=100)
```
